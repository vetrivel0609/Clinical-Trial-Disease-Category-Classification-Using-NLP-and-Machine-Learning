```python
# IMPORT LIBRARIES AND SET PROJECT PATHS

import os
import pandas as pd
import joblib
from scipy.sparse import load_npz
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

BASE_DIR = r"C:\Users\acer\Documents\Clinical_Trial_Classification"
OUTPUT_DIR = os.path.join(BASE_DIR, "outputs")
MODEL_DIR = os.path.join(BASE_DIR, "models")

print("Libraries imported successfully.")
print("Test features exist:",
      os.path.exists(os.path.join(OUTPUT_DIR, "X_test_tfidf.npz")))
print("Test labels exist:",
      os.path.exists(os.path.join(OUTPUT_DIR, "y_test.csv")))
print("Logistic Regression model exists:",
      os.path.exists(os.path.join(MODEL_DIR, "logistic_regression_model.pkl")))
print("Linear SVM model exists:",
      os.path.exists(os.path.join(MODEL_DIR, "linear_svm_model.pkl")))
```

```python
# LOAD TEST DATA AND TRAINED MODELS

X_test = load_npz(
    os.path.join(OUTPUT_DIR, "X_test_tfidf.npz")
)

y_test = pd.read_csv(
    os.path.join(OUTPUT_DIR, "y_test.csv")
)["target"]

logistic_model = joblib.load(
    os.path.join(MODEL_DIR, "logistic_regression_model.pkl")
)

svm_model = joblib.load(
    os.path.join(MODEL_DIR, "linear_svm_model.pkl")
)

print("Test data and models loaded successfully.")
print("Test feature matrix:", X_test.shape)
print("Test target shape:", y_test.shape)
print("Missing target values:", y_test.isna().sum())
print("Number of Logistic Regression classes:",
      len(logistic_model.classes_))
print("Number of Linear SVM classes:",
      len(svm_model.classes_))
```

```python
# GENERATE MODEL PREDICTIONS

y_pred_logistic = logistic_model.predict(X_test)
y_pred_svm = svm_model.predict(X_test)

print("Predictions generated successfully.")
print("Logistic Regression predictions:", len(y_pred_logistic))
print("Linear SVM predictions:", len(y_pred_svm))
```

```python
# CALCULATE MODEL EVALUATION METRICS

results = []

for model_name, y_pred in [
    ("Logistic Regression", y_pred_logistic),
    ("Linear SVM", y_pred_svm)
]:
    results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, average="weighted", zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, average="weighted", zero_division=0
        ),
        "F1_Score": f1_score(
            y_test, y_pred, average="weighted", zero_division=0
        )
    })

evaluation_results = pd.DataFrame(results)

print("MODEL EVALUATION RESULTS")
display(evaluation_results.round(4))
```

```python
# DETAILED CLASSIFICATION REPORTS

logistic_report = classification_report(
    y_test,
    y_pred_logistic,
    output_dict=True,
    zero_division=0
)

svm_report = classification_report(
    y_test,
    y_pred_svm,
    output_dict=True,
    zero_division=0
)

logistic_report_df = pd.DataFrame(logistic_report).transpose()
svm_report_df = pd.DataFrame(svm_report).transpose()

print("LOGISTIC REGRESSION CLASSIFICATION REPORT")
display(logistic_report_df.round(4))

print("LINEAR SVM CLASSIFICATION REPORT")
display(svm_report_df.round(4))
```

```python
# GENERATE CONFUSION MATRICES

logistic_cm = confusion_matrix(y_test, y_pred_logistic)
svm_cm = confusion_matrix(y_test, y_pred_svm)

print("LOGISTIC REGRESSION CONFUSION MATRIX")
display(pd.DataFrame(logistic_cm))

print("LINEAR SVM CONFUSION MATRIX")
display(pd.DataFrame(svm_cm))
```

```python
# MAP ENCODED LABELS TO DISEASE NAMES

label_encoder = joblib.load(
    os.path.join(MODEL_DIR, "label_encoder.pkl")
)

disease_names = label_encoder.inverse_transform(
    list(range(len(label_encoder.classes_)))
)

label_mapping = pd.DataFrame({
    "Encoded_Label": range(len(disease_names)),
    "Disease_Category": disease_names
})

display(label_mapping)
```

```python
# CONFUSION MATRICES WITH DISEASE NAMES

display_names = label_encoder.classes_

logistic_cm_df = pd.DataFrame(
    logistic_cm,
    index=display_names,
    columns=display_names
)

svm_cm_df = pd.DataFrame(
    svm_cm,
    index=display_names,
    columns=display_names
)

print("Logistic Regression - Confusion Matrix")
display(logistic_cm_df)

print("Linear SVM - Confusion Matrix")
display(svm_cm_df)
```

```python
# CELL 9: SAVE MODEL EVALUATION RESULTS

evaluation_results.to_csv(
    os.path.join(OUTPUT_DIR, "model_evaluation_summary.csv"),
    index=False
)

logistic_report_df.to_csv(
    os.path.join(OUTPUT_DIR, "logistic_regression_classification_report.csv")
)

svm_report_df.to_csv(
    os.path.join(OUTPUT_DIR, "linear_svm_classification_report.csv")
)

logistic_cm_df.to_csv(
    os.path.join(OUTPUT_DIR, "logistic_regression_confusion_matrix.csv")
)

svm_cm_df.to_csv(
    os.path.join(OUTPUT_DIR, "linear_svm_confusion_matrix.csv")
)

print("Evaluation results saved successfully.")
print("Saved files:")
print("1. model_evaluation_summary.csv")
print("2. logistic_regression_classification_report.csv")
print("3. linear_svm_classification_report.csv")
print("4. logistic_regression_confusion_matrix.csv")
print("5. linear_svm_confusion_matrix.csv")
```

```python
# LOAD TRAINING DATA FOR ACCURACY CHECK

X_train = load_npz(
    os.path.join(OUTPUT_DIR, "X_train_tfidf.npz")
)

y_train = pd.read_csv(
    os.path.join(OUTPUT_DIR, "y_train.csv")
)["target"]

print("Training features shape:", X_train.shape)
print("Training labels shape:", y_train.shape)
print("Missing training labels:", y_train.isna().sum())
```

```python
# TRAIN VS TEST ACCURACY

from sklearn.metrics import accuracy_score

train_results = []

for model_name, model in [
    ("Logistic Regression", logistic_model),
    ("Linear SVM", svm_model)
]:
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)

    train_results.append({
        "Model": model_name,
        "Train_Accuracy": train_accuracy,
        "Test_Accuracy": test_accuracy,
        "Accuracy_Gap": train_accuracy - test_accuracy
    })

train_test_results = pd.DataFrame(train_results)

display(train_test_results.round(4))
```

```python
# TRAIN VS TEST F1-SCORE

from sklearn.metrics import f1_score

f1_results = []

for model_name, model in [
    ("Logistic Regression", logistic_model),
    ("Linear SVM", svm_model)
]:
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    train_f1 = f1_score(
        y_train, y_train_pred,
        average="weighted",
        zero_division=0
    )

    test_f1 = f1_score(
        y_test, y_test_pred,
        average="weighted",
        zero_division=0
    )

    f1_results.append({
        "Model": model_name,
        "Train_F1_Score": train_f1,
        "Test_F1_Score": test_f1,
        "F1_Gap": train_f1 - test_f1
    })

f1_comparison = pd.DataFrame(f1_results)
display(f1_comparison.round(4))
```

```python
# CELL 12: FINAL MODEL SELECTION

final_comparison = evaluation_results.merge(
    train_test_results,
    on="Model"
).merge(
    f1_comparison,
    on="Model"
)

final_comparison = final_comparison.round(4)

final_comparison.to_csv(
    os.path.join(OUTPUT_DIR, "model_comparison.csv"),
    index=False
)

best_model = final_comparison.loc[
    final_comparison["Test_F1_Score"].idxmax()
]

print("FINAL MODEL COMPARISON")
display(final_comparison)

print("\nBEST MODEL")
print(f"Model            : {best_model['Model']}")
print(f"Test Accuracy    : {best_model['Test_Accuracy']:.4f}")
print(f"Test F1-Score    : {best_model['Test_F1_Score']:.4f}")
print(f"Accuracy Gap     : {best_model['Accuracy_Gap']:.4f}")
print(f"F1 Gap           : {best_model['F1_Gap']:.4f}")
print("\nModel comparison saved as: model_comparison.csv")
```
